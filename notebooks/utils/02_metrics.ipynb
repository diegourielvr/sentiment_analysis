# Metricas de evaluación

Se muestran métricas de los modelos que tuvieron mejor desempeño para cada dataset.

```python
import sys
import os

# Agregar el directorio raiz al PYTHONPATH
module_path = os.path.abspath(os.path.join('../..'))
if module_path not in sys.path:
    sys.path.append(module_path)
```

```python
import pandas as pd
from constants.constants_tiktok import METRICS_DIR as TIKTOK_METRICS
from constants.constants_twitter import METRICS_DIR as TWITTER_METRICS

def compare_metrics(dir: str):
    df_best = []
    for filename in os.listdir(dir):
        if filename.endswith(".csv"):
            # Leer csv
            df = pd.read_csv(os.path.join(dir, filename))
            df = df[['model', 'accuracy', 'precision', 'recall', 'f1_score', 'name', 'train_time']]
            df = df.loc[df['accuracy'].idxmax()]
            df_best.append(df)
    df_final = pd.DataFrame(df_best).reset_index(drop=True)
    df_final = df_final.sort_values(by='accuracy', ascending=False)
    return df_final

print("TIKTOK METRICS")
display(compare_metrics(TIKTOK_METRICS))
print("TWITTER METRICS")
display(compare_metrics(TWITTER_METRICS))
        
```
